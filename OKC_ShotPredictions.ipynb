Ensuring there are no anamolies between training and testing set other than outcome.

```python
import pandas as pd
import numpy as np

train = pd.read_csv("training.csv.gz", na_values=["NA"])
test  = pd.read_csv("testing.csv.gz",  na_values=["NA"])

print(train.shape, test.shape) # Check Dimensions of sets
print("In train but not test:", set(train.columns) - set(test.columns)) # Check column differences
print(train.isna().sum()[train.isna().sum() > 0]) # Shows columns with missing values and counts how are empty
```

Step 1 and 2 passed cleanly. Only outcome differs between train and test, and the missing counts match the contester counts exactly on every line. Missingness is fully structural, so we can confidently fill those later with a "no contester" value instead of imputing.

```python
print((train.shotclock == 24).sum(), (train.shotclock == 0).sum())
print((train.distance > 40).sum())
print(np.corrcoef(train.distance, np.hypot(train.locationx, train.locationy))[0, 1])
```

About 10,000 shots at exactly 24.0 and about 10,000 at 0.0. A real shot at 24.0 is nearly impossible, so this is probably the clock being switched off (end of a quarter when less than 24 seconds remain on the game clock). Those rows may need a flag so the model doesn't treat "clock off" as "brand new possession."

The coordinate check was the surprise. A correlation of -0.52 means the hoop is not at (0, 0).

```python
rim = train[train.distance < 2]
print(len(rim))
print(rim.locationx.round(-1).value_counts().sort_index())
print(rim.locationy.round(-1).value_counts().sort_index())
```

The hoop is at roughly x = -40, y = 0, and every rim shot is at the same end. So the data was already mirrored to one side for you, which saves a step. It also explains the -0.52 correlation: the origin is center court, the hoop sits about 42 feet down the x-axis, and x runs along the court's length while y runs across its width.

```python
diff = (both.distcont1 - both.closestdefdist).abs()
mism = both[diff > 0.01]
print("Rows that disagree:", len(mism))
print("Share where closest defender isn't contester 1:",
      round((mism.closestdef_id != mism.contester1_id).mean(), 3))
```

The distcont1 theory is confirmed: share = 1.0. In all 10,392 disagreeing rows, the closest defender was not the first contester. That's a meaningful situation (a defender is nearby but isn't challenging the shot, maybe trailing behind), so it becomes a feature instead of a cleaning problem.

```python
# Court setup:
# Define a function to draw the basketball court for future plots.
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle, Arc

def draw_court(ax, hx, color="black", lw=1.5):
    base = hx - 5.25                                   # baseline is 5.25 ft behind rim center
    parts = [
        Circle((0, hx), 0.75, fill=False),             # rim
        Rectangle((-3, hx - 1.25), 6, 0, fill=False),  # backboard
        Rectangle((-8, base), 16, 19, fill=False),     # paint
        Circle((0, base + 19), 6, fill=False),         # free throw circle
        Arc((0, hx), 8, 8, theta1=0, theta2=180),      # restricted area
        Arc((0, hx), 47.5, 47.5, theta1=22.1, theta2=157.9),  # 3pt arc
        Rectangle((-22, base), 0, 14.2, fill=False),   # corner 3 (left)
        Rectangle((22, base), 0, 14.2, fill=False),    # corner 3 (right)
        Rectangle((-25, base), 50, 47, fill=False),    # half court outline
    ]
    for p in parts:
        p.set_edgecolor(color); p.set_linewidth(lw); ax.add_patch(p)
    ax.set_xlim(-26, 26); ax.set_ylim(base - 2, base + 49)
    ax.set_aspect("equal"); ax.axis("off")
```

The following showcases the volume of feature specific shots represented as gray bars, and the make rate of those shots represented as the blue line.

```python
def rate_plot(df, col, bins=None, title=None):
    g = pd.cut(df[col], bins, include_lowest=True) if bins is not None else df[col]
    s = df.groupby(g, observed=True).outcome.agg(["mean", "size"])
    fig, ax1 = plt.subplots(figsize=(11, 4))
    ax1.bar(s.index.astype(str), s["size"], color="lightgray")
    ax1.set_ylabel("Shots")
    ax1.tick_params(axis="x", rotation=90)
    ax2 = ax1.twinx()
    ax2.plot(s.index.astype(str), s["mean"], marker="o", color="tab:blue")
    ax2.set_ylabel("Make rate"); ax2.set_ylim(0, 1)
    ax1.set_title(title or col)
    plt.tight_layout(); plt.show()
    return s

rate_plot(train, "distance", bins=np.arange(0, 48, 1))
rate_plot(train, "shotclock", bins=np.arange(0, 25, 1))
rate_plot(train, "shottype")
rate_plot(train, "num_contesters")

#shows volume as gray bars and make rate as a line on each feature
```

# Feature Prep

A note on choices ...

Distance capped at 35. The noisy heave tail we discussed. shottype == heave still tells the model those shots are special.

Defender distance clipped at 20, then log-transformed. Same shape as the distance cliff: a defender at 2 ft vs 4 ft matters a lot, 15 ft vs 17 ft barely matters. log1p compresses the far end so the model spends its effort where the differences are real.

Dribbles clipped at 15. A 36-dribble possession is an outlier that could pull a straight-line coefficient around.


```python
# =========== Original Feature Prep
def parse_ap(df):
    a = df.closestdefapproach.str.strip("{}").str.split(",", expand=True).astype(float)
    a.columns = [f"a{i}" for i in range(a.shape[1])]
    return a

def prep(df):
    out = pd.DataFrame(index=df.index)
    out["distance"]   = df.distance.clip(upper=35)
    out["shotclock"]  = df.shotclock
    out["defdist"]    = np.log1p(df.closestdefdist.clip(upper=20))
    out["dribbles"]   = df.dribblesbefore.clip(upper=15)
    out["speed"]      = df.shooterspeed
    out["contesters"] = df.num_contesters
    out["three"]      = df.three.astype(int)
    out["closest_not_contesting"] = ((df.num_contesters > 0) &
                                     (df.closestdef_id != df.contester1_id)).astype(int)
    a = parse_ap(df)
    out["def_delta"] = a.a3 - a.a0
    out["def_start"] = a.a0
    out["shottype"]  = df.shottype
    out["gamestate"] = df.gamestate

    out["angle"] = np.degrees(np.arctan2(np.abs(df.locationy - (-0.03)),
                                     np.abs(df.locationx - (-41.08))))
    out["corner3"] = (df.three & (np.abs(df.locationy) > 20)).astype(int)
    return out

X, y   = prep(train), train.outcome.astype(int)
X_test = prep(test)
```

```python
# ======= Feature Prep V2 ===================
def parse_ap(df):
    a = df.closestdefapproach.str.strip("{}").str.split(",", expand=True).astype(float)
    a.columns = [f"a{i}" for i in range(a.shape[1])]
    return a

def prep(df):
    out = pd.DataFrame(index=df.index)
    HX, HY = -41.08, -0.03

    # core
    out["distance"]   = df.distance.clip(upper=35)
    out["shotclock"]  = df.shotclock
    out["defdist"]    = np.log1p(df.closestdefdist.clip(upper=20))
    out["dribbles"]   = df.dribblesbefore.clip(upper=15)
    out["speed"]      = df.shooterspeed
    out["contesters"] = df.num_contesters
    out["three"]      = df.three.astype(int)
    out["closest_not_contesting"] = ((df.num_contesters > 0) &
                                     (df.closestdef_id != df.contester1_id)).astype(int)
    out["shottype"]   = df.shottype
    out["gamestate"]  = df.gamestate

    # BLOCK A: location
    out["abs_y"]   = (df.locationy - HY).abs()
    out["angle"]   = np.degrees(np.arctan2((df.locationy - HY).abs(),
                                           (df.locationx - HX).abs()))
    out["corner3"] = (df.three & ((df.locationy - HY).abs() > 20)).astype(int)

    # BLOCK B: defender geometry
    d = df[["distcont1","distcont2","distcont3","distcont4"]]
    out["cont_min"]   = d.min(axis=1)
    out["cont_mean"]  = d.mean(axis=1)
    out["cont_2nd"]   = df.distcont2
    out["n_within_4"] = (d < 4).sum(axis=1)
    out["n_within_6"] = (d < 6).sum(axis=1)

    a = parse_ap(df)
    out["def_start"] = a.a0
    out["def_delta"] = a.a3 - a.a0
    out["def_late"]  = a.a3 - a.a2
    out["def_accel"] = (a.a3 - a.a2) - (a.a1 - a.a0)
    out["close_rate_vs_speed"] = (a.a0 - a.a3) / (df.shooterspeed + 1)

    # BLOCK C: shot creation
    out["catch_shoot"] = (df.dribblesbefore == 0).astype(int)
    out["one_dribble"] = (df.dribblesbefore == 1).astype(int)
    return out
```

```python
# ===== prep v3 : complete =====

HX, HY = -41.08, -0.03

def parse_ap(df):
    a = df.closestdefapproach.str.strip("{}").str.split(",", expand=True).astype(float)
    a.columns = [f"a{i}" for i in range(a.shape[1])]
    return a

# shooter tables, built from TRAIN only
counts = train.shooter_id.value_counts()

prof = train.groupby("shooter_id").agg(
    three_rate=("three", "mean"),
    avg_dist=("distance", "mean"),
    n=("outcome", "size"))
prof.loc[prof.n < 100, ["three_rate", "avg_dist"]] = np.nan

K, lg = 200, train.outcome.mean()
made  = train.groupby("shooter_id").outcome.sum()
shots = train.groupby("shooter_id").outcome.size()
skill = ((made + K * lg) / (shots + K)) - lg


def prep(df):
    out = pd.DataFrame(index=df.index)

    # core
    out["distance"]   = df.distance.clip(upper=35)
    out["shotclock"]  = df.shotclock
    out["defdist"]    = np.log1p(df.closestdefdist.clip(upper=20))
    out["dribbles"]   = df.dribblesbefore.clip(upper=15)
    out["speed"]      = df.shooterspeed
    out["contesters"] = df.num_contesters
    out["three"]      = df.three.astype(int)
    out["closest_not_contesting"] = ((df.num_contesters > 0) &
                                     (df.closestdef_id != df.contester1_id)).astype(int)
    out["shottype"]   = df.shottype
    out["gamestate"]  = df.gamestate

    # A: location
    out["abs_y"]   = (df.locationy - HY).abs()
    out["angle"]   = np.degrees(np.arctan2((df.locationy - HY).abs(),
                                           (df.locationx - HX).abs()))
    out["corner3"] = (df.three & ((df.locationy - HY).abs() > 20)).astype(int)

    # B: defender geometry -- contester spacing
    d = df[["distcont1", "distcont2", "distcont3", "distcont4"]]
    out["cont_min"]   = d.min(axis=1)
    out["cont_mean"]  = d.mean(axis=1)
    out["cont_2nd"]   = df.distcont2
    out["n_within_4"] = (d < 4).sum(axis=1)
    out["n_within_6"] = (d < 6).sum(axis=1)

    # B: defender geometry -- closing motion
    a = parse_ap(df)
    out["def_start"] = a.a0
    out["def_delta"] = a.a3 - a.a0
    out["def_late"]  = a.a3 - a.a2
    out["def_accel"] = (a.a3 - a.a2) - (a.a1 - a.a0)
    out["close_rate_vs_speed"] = (a.a0 - a.a3) / (df.shooterspeed + 1)

    # Bx: dribble x defender interactions (must come AFTER def_delta / def_late exist)
    out["sep_created"]  = out["def_delta"] * (df.dribblesbefore > 0).astype(int)
    out["drib_x_close"] = df.dribblesbefore.clip(upper=6) * out["def_late"]

    # S: shooter
    out["shooter_vol"]   = df.shooter_id.map(counts).fillna(0)
    out["three_rate"]    = df.shooter_id.map(prof.three_rate)
    out["avg_dist"]      = df.shooter_id.map(prof.avg_dist)
    out["shooter_skill"] = df.shooter_id.map(skill)
    return out


X, y   = prep(train), train.outcome.astype(int)
X_test = prep(test)

base = ["distance", "shotclock", "defdist", "dribbles", "speed", "contesters",
        "three", "closest_not_contesting", "shottype", "gamestate"]
A   = ["abs_y", "angle", "corner3"]
B   = ["cont_min", "cont_mean", "cont_2nd", "n_within_4", "n_within_6",
       "def_start", "def_delta", "def_late", "def_accel", "close_rate_vs_speed"]
Bx  = B + ["sep_created", "drib_x_close"]
S1  = ["shooter_vol"]
S2  = ["shooter_vol", "three_rate", "avg_dist"]
S3  = ["shooter_vol", "three_rate", "avg_dist", "shooter_skill"]

print(X.shape)
```

# Logistic Regression

```python
# First predictive model:
# Build a logistic regression pipeline that preprocesses numerical
# and categorical features before estimating shot make probability.
# Numerical variables are standardized, categorical variables are
# one-hot encoded, and shot distance is modeled with spline terms
# to capture nonlinear relationships.
from sklearn.pipeline import make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, SplineTransformer
from sklearn.linear_model import LogisticRegression

num_cols = ["shotclock", "defdist", "dribbles", "speed", "contesters",
            "three", "closest_not_contesting"]
cat_cols = ["shottype", "gamestate"]

pre = ColumnTransformer([
    ("dist", SplineTransformer(n_knots=8, degree=3, knots="quantile"), ["distance"]),
    ("num",  StandardScaler(), num_cols),
    ("cat",  OneHotEncoder(handle_unknown="ignore"), cat_cols),
])
logit = make_pipeline(pre, LogisticRegression(max_iter=2000))
```

```python
scores = []
for tr, va in logo.split(X, groups=train.season_id):
    logit.fit(X.iloc[tr], y.iloc[tr]) # Training the model
    p = logit.predict_proba(X.iloc[va])[:, 1] # Generates predictions
    scores.append(log_loss(y.iloc[va], p)) # Computes performance
    print(train.season_id.iloc[va].iloc[0], "held out | log-loss:", round(scores[-1], 4)) # Compares against the baseline
print("Mean:", round(np.mean(scores), 4), "| constant baseline: 0.6897")
```

The following refits the model on all available data and extracts the coefficients so you can see which features have the strongest positive or negative influence on shot success:

```python
logit.fit(X, y)   # refit on both seasons for interpretation
pre_fit = logit.named_steps["columntransformer"]
names   = pre_fit.get_feature_names_out()
coefs   = logit.named_steps["logisticregression"].coef_[0]

s = pd.Series(coefs, index=names).drop(index=[n for n in names if n.startswith("dist__")])
print(s.reindex(s.abs().sort_values(ascending=False).index).round(3))
```

log-odds. Positive means the shot is more likely to go in, negative less likely. Because the numeric features were standardized, their sizes are directly comparable, so the largest absolute value is the strongest driver among them.

The Distance Curve

This holds every other feature fixed and sweeps distance across its range, so the line shows distance's effect in isolation.

```python
grid = X.iloc[:400].copy()
for c in num_cols: grid[c] = X[c].median()
grid["shottype"], grid["gamestate"] = "jumper", "HALFCOURT"
grid["distance"] = np.linspace(0, 35, 400)
grid["three"] = (grid.distance > 23.75).astype(int)

plt.plot(grid.distance, logit.predict_proba(grid)[:, 1])
plt.xlabel("Distance (ft)"); plt.ylabel("Predicted make probability")
plt.title("Distance effect, all else held at median"); plt.show()
```

## Calibration

```python
from sklearn.calibration import calibration_curve

oof = np.zeros(len(X))
for tr, va in logo.split(X, groups=train.season_id):
    logit.fit(X.iloc[tr], y.iloc[tr])
    oof[va] = logit.predict_proba(X.iloc[va])[:, 1]

tru, pred = calibration_curve(y, oof, n_bins=20, strategy="quantile")
plt.plot(pred, tru, marker="o"); plt.plot([0, 1], [0, 1], "k--")
plt.xlabel("Predicted probability"); plt.ylabel("Actual make rate")
plt.title("Calibration"); plt.show()
print("Prediction range:", round(oof.min(), 3), "to", round(oof.max(), 3))
```

```python
ap = X.join(train[["closestdefdist"]]) if False else None  # ignore
appr = train.closestdefapproach.str.strip("{}").str.split(",", expand=True).astype(float)
appr.columns = [f"a{i}" for i in range(appr.shape[1])]

print(appr.describe().round(2))
print(appr.corrwith(train.closestdefdist).round(3))
print("Monotonic decreasing rows:",
      round((appr.diff(axis=1).iloc[:, 1:] <= 0).all(axis=1).mean(), 3))
print(appr.join(train[["outcome"]]).corr().outcome.round(3))
```

The correlations with closestdefdist climb in a clean ladder: 0.618, 0.715, 0.841, 0.941. So a3 is the moment closest to release and a0 is the earliest. The means tell the same story: 8.38 → 7.42 → 6.33 → 5.58 feet. On average the defender closes about 2.8 feet during the window.

Monotonic Figure of 0.505, only half of the shots do the defenders close in steadily

Raw correlation with a binary outcome is almost always tiny, because a single shot is mostly luck. For comparison, distance is the strongest thing in this dataset and its outcome correlation is probably around -0.2. So -0.05 isn't as small as it looks.



```python
appr["delta"] = appr.a3 - appr.a0
appr["late"]  = appr.a3 - appr.a2
print(appr[["delta", "late"]].describe().round(2))
print(appr[["delta","late"]].join(train[["outcome"]]).corr().outcome.round(3))

b = pd.qcut(appr.delta, 10, duplicates="drop")
print(train.groupby(b, observed=True).outcome.agg(["mean","size"]).round(3))
```

The results above were to see make rate differences of large close-in and short close-in, where the idea is that a large close-in usually means the shooter had a wide open shot.
Shots where the defender closed the most (delta -22 to -8 ft) go in at 0.496, the highest bin. Shots where the defender moved away (+1.8 to +11.7) go in at 0.433, the lowest. Defender sprinting at you looks good for the shooter.

Hold release distance fixed and ask whether closing still matters:

```python
tmp = train[["outcome","closestdefdist"]].join(appr[["a0","delta","late"]])
print("corr(delta, a0):", round(tmp.delta.corr(tmp.a0), 3))

tmp["dist_bin"]  = pd.qcut(tmp.closestdefdist, 5)
tmp["delta_bin"] = pd.qcut(tmp.delta, 4, duplicates="drop")
print(tmp.pivot_table(index="dist_bin", columns="delta_bin",
                      values="outcome", aggfunc="mean", observed=True).round(3))
```

Reading rows Left to Right; Defender distance at release -> Closed Hard -> Closed Some -> Roughly steady -> Peeled off

Within a fixed defender distance, closing behavior still moves make rate by 10 to 14 points. The confounding is real (corr with a0 is -0.823) but it doesn't explain the whole effect. Your defender-intent reading survived the test.

And the pattern flips in the bottom two rows. When the defender is already 5+ feet away, closing hard slightly hurts the shooter. That makes sense as two different situations: at close range a hard closer is arriving late at a good look, but at long range a hard closer is a genuine contest on a jumper the shooter has time to see coming.

```python
# ===== ABLATION HELPER =====
# Same validation as the final model: leave-one-season-out, log-loss.
# Add one block of features at a time and see what each is actually worth.

def run(cols, label):
    Xa = X[cols].copy()
    for c in ["shottype", "gamestate"]:
        if c in Xa: Xa[c] = Xa[c].astype("category")
    s = []
    for tr, va in logo.split(Xa, groups=train.season_id):
        m = lgb.LGBMClassifier(**params)
        m.fit(Xa.iloc[tr], y.iloc[tr],
              eval_set=[(Xa.iloc[va], y.iloc[va])],
              eval_metric="binary_logloss",
              callbacks=[lgb.early_stopping(100, verbose=False)])
        s.append(log_loss(y.iloc[va], m.predict_proba(Xa.iloc[va])[:, 1]))
    print(f"{label:<28} {np.mean(s):.4f}")
    return np.mean(s)
```

```python
# ===== HYPERPARAMETER PASS =====
# Only knobs we've never touched. Same folds, same features, so any
# difference is the settings and nothing else.

def score_params(p, label):
    s = []
    for tr, va in logo.split(Xf, groups=train.season_id):
        m = lgb.LGBMClassifier(**p)
        m.fit(Xf.iloc[tr], y.iloc[tr],
              eval_set=[(Xf.iloc[va], y.iloc[va])],
              eval_metric="binary_logloss",
              callbacks=[lgb.early_stopping(100, verbose=False)])
        s.append(log_loss(y.iloc[va], m.predict_proba(Xf.iloc[va])[:, 1]))
    print(f"{label:<28} {np.mean(s):.4f}")
    return np.mean(s)


for lr, leaves in [(0.05, 63), (0.03, 63), (0.05, 127), (0.03, 31)]:
    score_params({**params, "learning_rate": lr, "num_leaves": leaves},
                 f"lr={lr}, leaves={leaves}")
```

```python
# effort premium: how much commitment exceeds what this shot type and distance gets on average
D["dbin2"] = pd.cut(D.distance, np.arange(0, 32, 3))
exp_commit = D.groupby(["shottype", "dbin2"], observed=True).committed.transform("mean")
D["over_commit"] = D.committed - exp_commit

j = D[D.shottype == "jumper"]
tab = j.groupby("abin", observed=True).agg(
    over_commit=("over_commit", "mean"),
    make=("outcome", "mean"),
    n=("outcome", "size"))
print("\nJumpers: commitment beyond what distance predicts")
print(tab.round(4))
```

```python
# ===== BEST SHOTS AGAINST A HARD CLOSEOUT =====
import numpy as np, pandas as pd, matplotlib.pyplot as plt

ap_t = test.closestdefapproach.str.strip("{}").str.split(",", expand=True).astype(float)
ap_t.columns = [f"a{i}" for i in range(ap_t.shape[1])]

T = test[["locationx","locationy","distance","shottype","gamestate",
          "closestdefdist","num_contesters","three"]].copy()
T["closed"] = ap_t.a0 - ap_t.a3
T["pred"]   = final.predict_proba(Xt)[:, 1]
T["exp_pts"] = T.pred * np.where(T.three, 3, 2)

# define "closing hard" generously: top third of closeouts
CUT = T.closed.quantile(0.67)
hard = T[T.closed >= CUT]
print(f"hard-closeout threshold: {CUT:.2f} ft covered in the last second")
print(f"shots meeting it: {len(hard):,} of {len(T):,}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6.5))

h1 = axes[0].hexbin(hard.locationy, hard.locationx, C=hard.pred,
                    reduce_C_function=np.mean, gridsize=38, cmap="RdYlGn", mincnt=25)
axes[0].set_title(f"Make probability under a hard closeout (>{CUT:.1f} ft closed)")

h2 = axes[1].hexbin(hard.locationy, hard.locationx, C=hard.exp_pts,
                    reduce_C_function=np.mean, gridsize=38, cmap="RdYlGn", mincnt=25)
axes[1].set_title("Expected points per shot, same shots")

for ax, h in zip(axes, [h1, h2]):
    ax.set_aspect("equal"); ax.axis("off")
    fig.colorbar(h, ax=ax, shrink=0.7)
plt.tight_layout(); plt.show()
```
